### Multi-Label Disease Co-Occurrence, Prevalence, and Patient Cohort Auditing

```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

BASE_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from src.config import TARGET_CLASSES, DATA_DIR
print(f"Target Pathology Classes: {TARGET_CLASSES}")
```

### Load Patient Cohort Metadata

```python
csv_path = DATA_DIR / 'synthetic_nih_cohort.csv'
if csv_path.exists():
    df = pd.read_csv(csv_path)
    print(f"Total Studies: {len(df)}, Unique Patients: {df['patient_id'].nunique()}")
    display(df.head())
else:
    print("Cohort CSV not found. Run scripts/train_model.py to initialize.")
```

### Disease Prevalence Distribution

```python
if csv_path.exists():
    counts = df[TARGET_CLASSES].sum()
    plt.figure(figsize=(8, 4))
    sns.barplot(x=counts.index, y=counts.values, palette='viridis')
    plt.title('Multi-Label Class Prevalence')
    plt.ylabel('Number of Positive Studies')
    plt.xticks(rotation=15)
    plt.tight_layout()
    plt.show()
```
