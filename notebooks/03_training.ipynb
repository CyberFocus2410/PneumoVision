```python
import sys
from pathlib import Path
import torch
import matplotlib.pyplot as plt

BASE_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from src.inference.engine import PneumoInferenceEngine
from src.config import SAMPLES_DIR

engine = PneumoInferenceEngine()
sample_path = SAMPLES_DIR / 'sample_pneumonia.png'
results = engine.predict(sample_path, save_heatmaps=True)

print(f"Primary Finding: {results['primary_finding']}")
for p in results['predictions']:
    print(f" - {p['label']:<18}: {p['probability_percent']:>5.1f}% (Threshold: {p['threshold']*100:.0f}%, {p['confidence_band']})")

# Show Grad-CAM Overlay
pna_cam = results['heatmaps'].get('Pneumonia')
if pna_cam:
    from PIL import Image
    img = Image.open(pna_cam['overlay_path'])
    plt.figure(figsize=(6, 6))
    plt.imshow(img)
    plt.title('Grad-CAM++ Spatial Attention Map: Pneumonia')
    plt.axis('off')
    plt.show()
```
