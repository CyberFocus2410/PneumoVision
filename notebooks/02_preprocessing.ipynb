# PneumoVision — 02: Preprocessing, CLAHE & Augmentation
### Contrast Limited Adaptive Histogram Equalization (CLAHE) & Patient Stratification

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

BASE_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from src.preprocessing.transforms import CLAHETransform, get_inference_transforms
from src.preprocessing.quality_check import assess_image_quality
from src.config import SAMPLES_DIR

sample_path = SAMPLES_DIR / 'sample_pneumonia.png'
if sample_path.exists():
    raw_img = Image.open(sample_path)
    clahe_op = CLAHETransform(clip_limit=2.0)
    enhanced_img = clahe_op(raw_img)
    
    is_acc, status, metrics = assess_image_quality(raw_img)
    print(f"Quality Status: {status}, Sharpness: {metrics.get('sharpness_index')}")
    
    fig, axes = plt.subplots(1, 2, figsize=(10, 5))
    axes[0].imshow(raw_img, cmap='gray')
    axes[0].set_title('Raw Radiograph')
    axes[0].axis('off')
    
    axes[1].imshow(enhanced_img)
    axes[1].set_title('CLAHE Contrast Enhanced')
    axes[1].axis('off')
    plt.show()
```
